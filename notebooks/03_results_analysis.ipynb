# 03 · Results analysis
Paired contrasts with source-clustered bootstrap CIs. These are the same numbers that `wm-eval report` writes.

```python
from pathlib import Path
import json
import pandas as pd

RESULTS = Path("../results")
# Pick a run explicitly, or use the most recent one that has trials.parquet.
RUN_ID = None
runs = sorted(p for p in RESULTS.iterdir() if (p / "trials.parquet").exists())
RUN = RESULTS / RUN_ID if RUN_ID else runs[-1]
print(RUN.name)
```

```python
from wm_eval.analysis.statistics import compare
df = pd.read_parquet(RUN / "trials.parquet")
cmp = compare(df)
cmp[cmp.metric.isin(["response_words", "task_success", "marker_present", "resp_vs_source_contradiction_rate",
                     "response_sentiment_negative", "paired_response_similarity"])]
```

```python
compare(df, group_by=["generator", "receiver"]).query("metric == 'task_success'")
```
